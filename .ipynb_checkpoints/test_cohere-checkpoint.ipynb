```python
from llama_index.llms.cohere import Cohere
from llama_index.embeddings.cohere import CohereEmbedding
from llama_index.core.llms import ChatMessage
from dotenv import load_dotenv
import os

# Load environment variables from .env file
load_dotenv()

# Initialize the LLM with updated model name
# Old "command-r-plus" was removed - use versioned model names
llm = Cohere(
    api_key=os.getenv("CO_API_KEY"),
    model="command-a-03-2025"  
)

# Use chat() method - requires a list of ChatMessage objects
messages = [
    ChatMessage(role="user", content="Hello! Can you hear me?")
]
response = llm.chat(messages)
print(response.message.content)
```
